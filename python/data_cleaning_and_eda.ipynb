#  EDA(Exploratory Data Analysis)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,8))
sns.set_style('whitegrid')
plt.show()
```

```python
df=pd.read_csv("/content/cleaned_hotel_bookings (1).csv")
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df['FINAL_REVENUE']=df['ADR']*df['TOTAL_NIGHTS']
```

```python
revenue_by_hotel=df.groupby('HOTEL')['FINAL_REVENUE'].sum().sort_values(ascending=False).astype(np.int32)
revenue_by_hotel
```

```python
df.rename(
    columns={'FINAL_REVENUE':'TOTAL_REVENUE'},
    inplace=True
)
```

```python
df.head()
```

```python
df.loc[ df['HOTEL'] == 'Resort Hotel', 'ADR'] *= 1.2
```

```python
df['REVENUE']=df['ADR']*df['TOTAL_NIGHTS']
```

```python

revenue_by_hotel.plot(kind='bar',color=['skyblue','pink'])
plt.title("Revenue by Hotel")
plt.xlabel("Hotel")
plt.ylabel("income")
```

```python
df.groupby("HOTEL")["REVENUE"].mean()
```

```python
from matplotlib.typing import MarkerType
monthly_revenue=df.groupby('MONTH')["REVENUE"].sum()
monthly_revenue.plot(kind="line",color="blue",marker='o')
plt.title("Monthly Revenue")
plt.xticks(rotation=90)
plt.ylabel("Revenue")
```

```python
top_country=df.groupby('COUNTRY')['REVENUE'].sum().sort_values(ascending=False)[:10]
top_country.plot(kind='bar',color='skyblue')
plt.title("Top 10 Countries by Revenue")
plt.xlabel("Country")
plt.ylabel("Revenue")
```

```python
top_country=df.groupby('COUNTRY')['FINAL REVENUE'].sum().sort_values(ascending=False)[-10:]
top_country.plot(kind='bar',color='skyblue')
plt.title("10 Countries by Lowest Revenue")
plt.xlabel("Country")
plt.ylabel("Revenue")
```

```python
sns.countplot(x='BOOKING_STATUS',data=df)
plt.title("Booking Status Distribution")
plt.xlabel("Booking Status")
plt.ylabel("Count")
plt.show()
```
